```

import pandas as pd
from google.colab import files

uploaded = files.upload()
file_name = next(iter(uploaded))

df = pd.read_csv(file_name, encoding="latin1")

print("Dataset loaded successfully!")
print("Rows and columns:", df.shape)
df.head()
```

```

# Count missing values in each column
missing_count = df.isnull().sum()

print("Missing values by column:")
print(missing_count)
```

```

summary = pd.DataFrame({
    "Column": df.columns,
    "Data Type": df.dtypes.astype(str).values,
    "Total Rows": len(df),
    "Non-Missing Count": df.notnull().sum().values,
    "Missing Count": df.isnull().sum().values,
    "Missing Percentage": (
        df.isnull().sum().values / len(df) * 100
    ).round(2)
})

summary = summary.sort_values(
    by="Missing Count",
    ascending=False
)

summary
```

```

total_missing = int(df.isnull().sum().sum())
columns_with_missing = int((df.isnull().sum() > 0).sum())
rows_with_missing = int(df.isnull().any(axis=1).sum())

print("Total missing cells:", total_missing)
print("Columns containing missing values:", columns_with_missing)
print("Rows containing at least one missing value:", rows_with_missing)

if total_missing == 0:
    print("No missing values detected.")
else:
    print("Missing values were detected. Review the summary before deciding how to handle them.")
```

```


output_file = "Task_12_Missing_Value_Summary.xlsx"

with pd.ExcelWriter(output_file, engine="openpyxl") as writer:
    summary.to_excel(
        writer,
        sheet_name="Missing Value Summary",
        index=False
    )

    df.to_excel(
        writer,
        sheet_name="Raw Data",
        index=False
    )

files.download(output_file)
```
